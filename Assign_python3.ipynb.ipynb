```python
#PART - A DICTIONARIES
#1. crate an empty Dictionary.
inventory = {}
print("Empty Dictionary: ")
print(inventory)
```

```python
#2.  Store the first product details in variable: Product Name = Mobile Phone Product Quantity = 5 Product Price = 20000 Product Release Year = 2020.
ProductNo1_Name = "Mobile Phone"
ProductNo1_Quantity = 5
ProductNo1_Price = 20000
ProductNo1_ReleaseYear = 2020

print("\nFirst Product Details:")
print("Product Name:", ProductNo1_Name)
print("Product Quantity:", ProductNo1_Quantity)
print("Product Price:", ProductNo1_Price)
print("Product Release Year:", ProductNo1_ReleaseYear)

```

```python
#3.add details in inventory
inventory["ProductNo1"] = {
    "Product Name": ProductNo1_Name,
    "Product Quantity": ProductNo1_Quantity,
    "Product Price": ProductNo1_Price,
    "Product Release Year": ProductNo1_ReleaseYear
}

print("\nFirst Product Added to Inventory:")
print(inventory)
```

```python
#4. Store the Second product details in variable.
ProductNo2_Name = "Laptop"
ProductNo2_Quantity = 3
ProductNo2_Price = 50000
ProductNo2_ReleaseYear = 2022

print("\nSecond Product Details:")
print("Product Name:", ProductNo2_Name)
print("Product Quantity:", ProductNo2_Quantity)
print("Product Price:", ProductNo2_Price)
print("Product Release Year:", ProductNo2_ReleaseYear)

```

```python
#5.Ass the item details in inventory
inventory["ProductNo2"] = {
    "Product Name": ProductNo2_Name,
    "Product Quantity": ProductNo2_Quantity,
    "Product Price": ProductNo2_Price,
    "Product Release Year": ProductNo2_ReleaseYear
}

print("\nSecond Product Added to Inventory:")
print(inventory)
```

```python
#6.Display the products present in inventory
print("\nProducts Present in Inventory:")

for product, details in inventory.items():
    print("\n", product)
    print("Product Name:", details["Product Name"])
    print("Product Quantity:", details["Product Quantity"])
    print("Product Price:", details["Product Price"])
    print("Product Release Year:", details["Product Release Year"])
```

```python
#7.Check if ProductNo1_releaseYear and ProductNo2_releaseYear are in inventory.
print("\nChecking Release Years in Inventory:")

if ProductNo1_ReleaseYear in inventory["ProductNo1"].values():
    print("ProductNo1_releaseYear is in inventory")
else:
    print("ProductNo1_releaseYear is not in inventory")

if ProductNo2_ReleaseYear in inventory["ProductNo2"].values():
    print("ProductNo2_releaseYear is in inventory")
else:
    print("ProductNo2_releaseYear is not in inventory")
```

```python
#8. Delete release year of both the products from the inventory.
del inventory["ProductNo1"]["Product Release Year"]
del inventory["ProductNo2"]["Product Release Year"]

print("\nInventory After Deleting Release Year:")
print(inventory)
```

```python
#PART - B TUPLES
#create a tuple called prices containing the following values:
prices = (250, 300, 150, 400, 100, 350, 200)
print("\nProduct Prices: ")
print(prices)
```

```python
#find and print the highest and lowest price.
highest_price = max(prices)
print("\nHiggest Price: ")
print(highest_price)

lowest_price = min(prices)
print("\nLowest Price: ")
print(lowest_price)
```

```python
#calculate and print the total sum of all prices.
total_sum = sum(prices)
print("\nTotal sum of all prices: ")
print(total_sum)
```

```python
#convert the tuple into sorted list(ascending order) and print it.
sorted_list = sorted(prices)
print("\nSorted Prices in ascending order: ")
print(sorted_list)
```
